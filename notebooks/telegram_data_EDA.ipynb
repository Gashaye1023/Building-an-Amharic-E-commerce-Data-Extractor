```python
def clean_amharic_text(text):
    if isinstance(text, str):  # Check if the input is a string
        text = re.sub(r'[^\u1200-\u137F\s]', '', text)  # Keep only Amharic characters and spaces
        text = re.sub(r'\s+', ' ', text)  # Normalize spaces
        return text.strip()
    return '' 
```

```python
def tokenize(text):
    return nltk.word_tokenize(text)  # Tokenize using NLTK
```

```python
import re

def simple_amharic_tokenizer(text):
    """A simple tokenizer for Amharic text"""
    # Split on whitespace and punctuation
    tokens = re.findall(r'[\w\u1200-\u137F]+|[\.,!?።፣፤፥፦፧፨]', text)
    return tokens

def interactive_labeling(messages):
    labeled_data = []
    
    for i, message in enumerate(messages):
        print(f"\nMessage {i+1}/{len(messages)}:")
        print(message)
        
        tokens = simple_amharic_tokenizer(message)
        tagged_tokens = []
        
        for token in tokens:
            print(f"\nToken: {token}")
            print("Available tags:")
            print("1. O (Outside)")
            print("2. B-PRODUCT (Beginning of product)")
            print("3. I-PRODUCT (Inside product)")
            print("4. B-LOC (Beginning of location)")
            print("5. I-LOC (Inside location)")
            print("6. B-PRICE (Beginning of price)")
            print("7. I-PRICE (Inside price)")
            
            while True:
                try:
                    choice = input("Select tag (1-7) or 's' to skip: ").strip().lower()
                    if choice == 's':
                        tag = 'O'
                        break
                    choice = int(choice)
                    if 1 <= choice <= 7:
                        break
                    print("Please enter a number between 1 and 7")
                except ValueError:
                    print("Please enter a valid number")
            
            if choice != 's':
                tags = ['O', 'B-PRODUCT', 'I-PRODUCT', 'B-LOC', 'I-LOC', 'B-PRICE', 'I-PRICE']
                tag = tags[choice-1]
            tagged_tokens.append((token, tag))
        
        labeled_data.append(tagged_tokens)
        print("\nCurrent tagging:")
        for token, tag in tagged_tokens:
            print(f"{token}\t{tag}")
        
        cont = input("\nContinue to next message? (y/n): ").lower()
        if cont != 'y':
            break
    
    return labeled_data

def save_conll(labeled_data, filename):
    with open(filename, 'w', encoding='utf-8') as f:
        for sentence in labeled_data:
            for token, tag in sentence:
                f.write(f"{token}\t{tag}\n")
            f.write("\n")

messages = df1['Cleaned_Message'].tolist()


print("Starting interactive labeling...")
labeled_data = interactive_labeling(messages[:50])  # Label first 50 messages
save_conll(labeled_data, 'manually_labeled_amharic.conll')
print("\nLabeling complete. Saved to 'manually_labeled_amharic.conll'")
```

```python
# Define entity labeling function
def label_entities(message):
    tokens = message.split()
    labels = []
    
    for token in tokens:
        # Labeling based on simple rules for demonstration
        if 'ዋጋ' in token or 'ብር' in token:
            labels.append('B-PRICE' if 'ዋጋ' in token else 'I-PRICE')
        elif token.lower() in ["addis", "abeba", "bole"]:
            labels.append('B-LOC' if token.lower() == "addis" else 'I-LOC')
        elif "Baby" in token:
            labels.append('B-Product')
        else:
            labels.append('O')
    
    return list(zip(tokens, labels))

# Prepare CoNLL format
def prepare_conll_format(df):
    conll_lines = []
    for _, row in df.iterrows():
        message = row['Cleaned_Message']
        labeled_entities = label_entities(message)
        
        for token, label in labeled_entities:
            conll_lines.append(f"{token} {label}")
        
        conll_lines.append("")  # Blank line to separate messages

    return "\n".join(conll_lines)

# Generate CoNLL formatted output
conll_output = prepare_conll_format(df1)

# Save to a plain text file
with open('labeled_data.conll', 'w', encoding='utf-8') as f:
    f.write(conll_output)

print("Labeled data saved to 'labeled_data.conll'")
```
